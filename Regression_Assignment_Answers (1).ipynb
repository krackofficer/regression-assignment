# Regression Assignment Answers

```python
# 1. What is Simple Linear Regression
print('''Simple Linear Regression is a statistical method that models the relationship between a dependent variable and a single independent variable using a straight line (Y = mX + c).''')
```

```python
# 2. What are the key assumptions of Simple Linear Regression
print('''The key assumptions include linearity, independence of errors, homoscedasticity, normal distribution of errors, and no multicollinearity.''')
```

```python
# 3. What does the coefficient m represent in the equation Y=mX+c
print('''The coefficient m represents the slope of the regression line, indicating how much Y changes for a one-unit change in X.''')
```

```python
# 4. What does the intercept c represent in the equation Y=mX+c
print('''The intercept c represents the value of Y when X is 0; it's the point where the regression line crosses the Y-axis.''')
```

```python
# 5. How do we calculate the slope m in Simple Linear Regression
print('''The slope m is calculated using the formula: m = Σ((X - X̄)(Y - Ȳ)) / Σ((X - X̄)^2).''')
```

```python
# 6. What is the purpose of the least squares method in Simple Linear Regression
print('''The least squares method minimizes the sum of the squares of the residuals to find the best-fitting line.''')
```

```python
# 7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression
print('''R² measures the proportion of variance in the dependent variable that is predictable from the independent variable(s).''')
```

```python
# 8. What is Multiple Linear Regression
print('''Multiple Linear Regression models the relationship between a dependent variable and two or more independent variables.''')
```

```python
# 9. What is the main difference between Simple and Multiple Linear Regression
print('''The main difference is that Simple Linear Regression uses one independent variable, while Multiple uses two or more.''')
```

```python
# 10. What are the key assumptions of Multiple Linear Regression
print('''Assumptions include linearity, independence, homoscedasticity, normal distribution of residuals, and no multicollinearity.''')
```

```python
# 11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model
print('''Heteroscedasticity refers to non-constant variance of residuals, which can lead to inefficient estimates and incorrect inference.''')
```

```python
# 12. How can you improve a Multiple Linear Regression model with high multicollinearity
print('''By removing or combining correlated predictors, or using techniques like Ridge or Lasso regression.''')
```

```python
# 13. What are some common techniques for transforming categorical variables for use in regression models
print('''One-hot encoding and label encoding are commonly used to convert categorical variables into numerical format.''')
```

```python
# 14. What is the role of interaction terms in Multiple Linear Regression
print('''Interaction terms model the combined effect of two or more variables, capturing relationships not seen with main effects alone.''')
```

```python
# 15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression
print('''In Simple Linear Regression, the intercept is the value of Y when X=0; in Multiple, it's when all predictors are zero.''')
```

```python
# 16. What is the significance of the slope in regression analysis, and how does it affect predictions
print('''The slope shows the effect of a one-unit increase in the predictor on the dependent variable, holding others constant.''')
```

```python
# 17. How does the intercept in a regression model provide context for the relationship between variables
print('''It helps understand the baseline value of Y and provides context when X=0 or all predictors are zero.''')
```

```python
# 18. What are the limitations of using R² as a sole measure of model performance
print('''R² doesn't account for the number of predictors and can be misleading; adjusted R² is more reliable.''')
```

```python
# 19. How would you interpret a large standard error for a regression coefficient
print('''A large standard error suggests the coefficient estimate is imprecise and may not be statistically significant.''')
```

```python
# 20. How can heteroscedasticity be identified in residual plots, and why is it important to address it
print('''It appears as a funnel shape in residual plots; addressing it ensures valid statistical inference.''')
```

```python
# 21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²
print('''It indicates that adding predictors might not be contributing meaningfully to the model.''')
```

```python
# 22. Why is it important to scale variables in Multiple Linear Regression
print('''To ensure that variables contribute proportionally and to help with model convergence.''')
```

```python
# 23. What is polynomial regression
print('''Polynomial regression models the relationship by using polynomial terms of the independent variable(s).''')
```

```python
# 24. How does polynomial regression differ from linear regression
print('''Polynomial regression includes higher-degree terms, allowing for curved lines instead of straight ones.''')
```

```python
# 25. When is polynomial regression used
print('''It is used when data shows a non-linear relationship that a linear model cannot capture.''')
```

```python
# 26. What is the general equation for polynomial regression
print('''The general equation is Y = b0 + b1X + b2X² + ... + bnXⁿ.''')
```

```python
# 27. Can polynomial regression be applied to multiple variables
print('''Yes, by including interaction and polynomial terms of multiple variables.''')
```

```python
# 28. What are the limitations of polynomial regression
print('''It can overfit the data, be sensitive to outliers, and may become complex with high degrees.''')
```

```python
# 29. What methods can be used to evaluate model fit when selecting the degree of a polynomial
print('''Cross-validation and information criteria (AIC, BIC) are used to select the optimal degree.''')
```

```python
# 30. Why is visualization important in polynomial regression
print('''Visualization helps identify non-linear patterns and assess model fit intuitively.''')
```

```python
# 31. How is polynomial regression implemented in Python?
print('''By using `PolynomialFeatures` from `sklearn.preprocessing` along with a linear model like `LinearRegression`.''')
```
